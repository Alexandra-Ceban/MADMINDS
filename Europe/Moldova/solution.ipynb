```python
import itertools
import pandas as pd

register_df = pd.read_csv('ring_register.csv')
observations_df = pd.read_csv('observations.csv')
```

```python
# make string
register_codes = set(
    register_df.iloc[:, 0].astype(str).str.upper().str.strip()
)
observations = observations_df.iloc[:, 12].astype(str).str.upper().str.strip()
```

```python
import csv

# din registru
register_codes = set()
with open("ring_register.csv", mode="r", encoding="utf-8") as f:
  reader = csv.reader(f)
  for row in reader:
    if row:
      code = row[0].strip().upper()
      if len(code) == 6:
        register_codes.add(code)

# din observatii
observations = []
with open("observations.csv", mode="r", encoding="utf-8") as f:
  reader = csv.reader(f)
  for row in reader:
    if row:
      obs = row[12].strip().upper()
      if obs and obs != "NAN":
        observations.append(obs)


alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
found_bird = None

for obs in observations:

  if obs not in register_codes and len(obs) == 6:
    print(f"Am găsit observația suspectă: {obs}")

  
    if "?" in obs:
      pos = obs.index("?")
      for letter in alphabet:
        candidate = obs[:pos] + letter + obs[pos + 1 :]
        if candidate in register_codes:
          found_bird = candidate
          break

    else:
      best_match = None
      min_diffs = 999
      for reg in register_codes:
        diffs = sum(c1 != c2 for c1, c2 in zip(obs, reg))
        if diffs < min_diffs:
          min_diffs = diffs
          best_match = reg
      found_bird = best_match

    break


print(f"\nInelul corect (WORD): {found_bird}")
print(f"FLAG-ul final este: SIGMOID_{found_bird} : {found_bird}")
```
